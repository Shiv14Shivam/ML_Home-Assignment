```python
import joblib

from sklearn.linear_model import LinearRegression
```

```python


X_train = joblib.load("X_train.pkl")
X_val = joblib.load("X_val.pkl")
X_test = joblib.load("X_test.pkl")

y_train = joblib.load("y_train.pkl")
y_val = joblib.load("y_val.pkl")
y_test = joblib.load("y_test.pkl")

X_train_scaled = joblib.load("X_train_scaled.pkl")
X_val_scaled = joblib.load("X_val_scaled.pkl")
X_test_scaled = joblib.load("X_test_scaled.pkl")

scaler = joblib.load("scaler.pkl")

print("Data loaded successfully")
```

```python


simple_model = LinearRegression()

simple_model.fit(X_train[["MedInc"]], y_train)

val_pred = simple_model.predict(X_val[["MedInc"]])
```

```python
from sklearn.metrics import mean_squared_error, r2_score

print("Validation MSE:", mean_squared_error(y_val, val_pred))
print("Validation R2:", r2_score(y_val, val_pred))
```

```python
multi_model = LinearRegression()

multi_model.fit(X_train, y_train)
import joblib

joblib.dump(multi_model, "model.pkl")

print("Model saved successfully!")

val_pred_multi = multi_model.predict(X_val)

print("Validation MSE:", mean_squared_error(y_val, val_pred_multi))
print("Validation R2:", r2_score(y_val, val_pred_multi))
```

```python
test_pred = multi_model.predict(X_test)

print("Test MSE:", mean_squared_error(y_test, test_pred))
print("Test R2:", r2_score(y_test, test_pred))
```

```python
import matplotlib.pyplot as plt

plt.scatter(y_test, test_pred)
plt.xlabel("Actual Values")
plt.ylabel("Predicted Values")
plt.title("Actual vs Predicted House Values")
plt.show()
```
